# Rename columns

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
c2023_csv = '../data/custom/original/c2023.csv'
c2022_csv = '../data/custom/original/c2022.csv'
c2021_csv = '../data/custom/original/c2021.csv'
c2020_csv = '../data/custom/original/c2020.csv'

df23 = pd.read_csv(c2023_csv)
df22 = pd.read_csv(c2022_csv)
df21 = pd.read_csv(c2021_csv)
df20 = pd.read_csv(c2020_csv)

# Remove 'DRVGR2023.' prefix from column names
print("original column names:", df23.columns.tolist())
print("original column names:", df22.columns.tolist())
print("original column names:", df21.columns.tolist())
print("original column names:", df20.columns.tolist())

df23.columns = df23.columns.str.replace('DRVGR2023.', '').str.replace('DRVEF2023.', '')
df23.head()
```

```python
print(df23.columns)
# Set all dataframes to use the same column names as df23
df22.columns = df23.columns
df21.columns = df23.columns 
df20.columns = df23.columns
```

```python
df22.head()
```

```python
df21.shape
```

```python
# Join all dataframes, ensuring columns match
df_all = pd.concat([df20, df21, df22, df23], axis=0, join='inner', ignore_index=True)

# Convert year column to string
df_all['year'] = df_all['year'].astype(str)

# Check which columns have quotes in their names
quoted_cols = [col for col in df_all.columns if '"' in col]
print("Columns with quotes:")
for col in quoted_cols:
    print(f"  {col}")
    
# Remove quotes from column names
df_all.columns = df_all.columns.str.replace('"', '')

# Verify no more quoted columns
quoted_cols = [col for col in df_all.columns if '"' in col]
print("\nRemaining quoted columns:", quoted_cols)


print("Original shapes:", df20.shape, df21.shape, df22.shape, df23.shape)
print("Joined shape:", df_all.shape)
print("Columns dropped:", set(df23.columns) - set(df_all.columns))
df_all.to_csv('../data/custom/processed/c20-23.csv', index=False)
```

```python
df23.to_csv('../data/custom/processed/c2023.csv', index=False)
df22.to_csv('../data/custom/processed/c2022.csv', index=False)
df21.to_csv('../data/custom/processed/c2021.csv', index=False)
df20.to_csv('../data/custom/processed/c2020.csv', index=False)
```

```python
df_csv = '../data/custom/processed/c20-23.csv'
df = pd.read_csv(df_csv)
df.head()
```

```python
print("\nUnique years in dataset:")
print(df['year'].unique())
print("Columns:")
for col in df.columns:
    print(col)
```

```python
# Get graduation rates for NJIT
njit_grad_rates = df[
    df['institution name'] == 'New Jersey Institute of Technology'
][
    ['institution name', 'year', 
    'Graduation rate, total cohort', 
    'Graduation rate - Bachelor degree within 4 years, total',
    'Graduation rate - Bachelor degree within 5 years, total',
    'Graduation rate - Bachelor degree within 6 years, total']
]
njit_grad_rates.head()
```

```python
# Plot graduation rates over time
plt.figure(figsize=(10,6))

# Create arrays for each metric
years = njit_grad_rates['year'].values - 6  # Subtract 6 from years for cohort year
total_rates = njit_grad_rates['Graduation rate, total cohort'].values
four_year = njit_grad_rates['Graduation rate - Bachelor degree within 4 years, total'].values
five_year = njit_grad_rates['Graduation rate - Bachelor degree within 5 years, total'].values
six_year = njit_grad_rates['Graduation rate - Bachelor degree within 6 years, total'].values

# Plot connected lines with markers for each metric
plt.plot(years, total_rates, marker='o', label='Total')
plt.plot(years, four_year, marker='s', label='4-year') 
plt.plot(years, five_year, marker='^', label='5-year')
plt.plot(years, six_year, marker='D', label='6-year')

# Add text labels for each point
for i, year in enumerate(years):
    plt.text(year, total_rates[i], f"{total_rates[i]}%", verticalalignment='bottom')
    plt.text(year, four_year[i], f"{four_year[i]}%", verticalalignment='bottom')
    plt.text(year, five_year[i], f"{five_year[i]}%", verticalalignment='bottom')
    plt.text(year, six_year[i], f"{six_year[i]}%", verticalalignment='bottom')

plt.title('NJIT Graduation Rates by Cohort Year')
plt.xlabel('Cohort Year') 
plt.ylabel('Graduation Rate (%)')
plt.legend()
plt.xticks(years)
plt.show()
```

```python
# Default to NJIT if no institution is selected
selected_institution = 'New Jersey Institute of Technology'

# Filter data for the selected institution
inst_df = df[df['institution name'] == selected_institution].copy()
inst_df.head()

if inst_df.empty:
    raise ValueError(f"No data found for institution: {selected_institution}")

# Compute cohort year (subtract 6)
inst_df['Cohort Year'] = inst_df['year'] - 6

# Melt the data for easier plotting
plot_df = inst_df.melt(
    id_vars=['Cohort Year', 'institution name'],
    value_vars=[
        'Graduation rate, total cohort',
        'Graduation rate - Bachelor degree within 4 years, total',
        'Graduation rate - Bachelor degree within 5 years, total',
        'Graduation rate - Bachelor degree within 6 years, total'
    ],
    var_name='Metric',
    value_name='Graduation Rate (%)'
)

# Simplify the metric labels for cleaner legend
label_map = {
    'Graduation rate, total cohort': 'Total Cohort',
    'Graduation rate - Bachelor degree within 4 years, total': '4-Year',
    'Graduation rate - Bachelor degree within 5 years, total': '5-Year',
    'Graduation rate - Bachelor degree within 6 years, total': '6-Year'
}
plot_df['Metric'] = plot_df['Metric'].map(label_map)

# Plot with Plotly Express
fig = px.line(
    plot_df,
    x='Cohort Year',
    y='Graduation Rate (%)',
    color='Metric',
    markers=True,
    title=f"{selected_institution} Graduation Rates by Cohort Year",
    hover_data={'Graduation Rate (%)': ':.1f'}
)

fig.update_traces(text=plot_df['Graduation Rate (%)'].round(1), textposition="top center")
fig.update_layout(
    xaxis=dict(title='Cohort Year', tickmode='linear'),
    yaxis_title='Graduation Rate (%)',
    legend_title='Rate Type',
    template='plotly_white',
    title_x=0.5
)
```

# Map states 

```python
df_csv = '../data/custom/processed/c20-23.csv'
dfcsv = pd.read_csv(df_csv)
dfcsv.head()
```

```python
df_info = '../data/schoolinfo.csv'
dfinfo = pd.read_csv(df_info)
dfinfo.head()
```

```python
# Check null rows in dfcsv
print("Null values in custom dataset (dfcsv):")
print(dfcsv.isnull().sum().sort_values(ascending=False))

print("\nNull values in school info dataset (dfinfo):")
print(dfinfo.isnull().sum().sort_values(ascending=False))
```

```python
# Merge dfinfo into dfcsv based on unitid
dfcsv = dfcsv.merge(
    dfinfo[['UNITID', 'STATE', 'STABBR']], 
    left_on='unitid',
    right_on='UNITID',
    how='left'
)

# Drop duplicate UNITID column from merge
dfcsv = dfcsv.drop('UNITID', axis=1)
dfcsv.head()
```

```python
# Save merged dataset back to csv
output_path = '../data/custom/processed/custom.csv'
dfcsv.to_csv(output_path, index=False)
```
